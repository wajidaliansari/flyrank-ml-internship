## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd

# Load the FlyRank starter dataset
# (If you get a 'file not found' error, change the path to just 'content_refresh_anonymized.csv')
file_path = 'content_refresh_anonymized.csv'

df = pd.read_csv(file_path)

# Filter for valid pages (pages with search demand that are older than 90 days)
df_valid = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)].copy()

# Calculate the real numbers to justify the project
total_valid_pages = len(df_valid)
declining_pages = (df_valid['trend_direction'] == 'down').sum()

# Print the 2-3 real numbers requested by the assignment
print(f"1. Total eligible mature pages analyzed: {total_valid_pages:,}")
print(f"2. Pages actively trending down in visibility: {declining_pages:,}")
print("Conclusion: The high volume of decaying pages proves the need for a ranked priority score.")
```
